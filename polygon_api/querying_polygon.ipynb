```python
from massive import RESTClient
import pandas as pd
```

```python
with open("key.txt") as file:
    API_key = file.read()

API_key = API_key.strip()
```

```python
client = RESTClient(api_key=API_key)
```

```python
def query_ohlcv(ticker,multiplier=1, timespan="day", start="2024-05-08", end="2026-05-07",adjusted=True):
    # List Aggregates (Bars)
    aggs = []
    for a in client.list_aggs(ticker=ticker, multiplier=1, timespan=timespan, from_=start, to=end,adjusted=adjusted):
        aggs.append(a)
    df = pd.DataFrame(aggs)
    df['date'] = pd.to_datetime(df['timestamp'], unit='ms')
    df.to_csv(f"./ohlcv/{ticker}.csv")
    return df
    
```

```python
ticker = "AAPL"
df = query_ohlcv(ticker=ticker, multiplier=1, timespan="day", start="2024-05-08", end="2026-05-07",adjusted=True)
df.head()
```

```python
new_df = df.copy()
```

```python
new_df.columns = new_df.columns.droplevel(1)
```

```python
# Reset index to move 'Date' from the index to a column
new_df = new_df.rename_axis(None, axis=1)
new_df = new_df.reset_index()
new_df.head()
```

```python
new_df.tail()
```

```python
new_df['year'] = new_df['Date'].dt.year
```

```python
new_new_df = new_df.query("year >= 2000")
```

```python
new_new_df['stock_id'] = 1
```

```python
new_new_df = new_new_df.rename(columns ={
                    "Date":"date",
                    "Open":"open",
                    "Close":"close",
                    "Low":"low",
                    "High":"high",
                    "Volume":"volume"
                    })
```
